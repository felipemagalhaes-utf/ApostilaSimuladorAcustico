# Soluções dos Exercícios de Ray-Tracing e Propagação

Para a resolução dos exercícios deste capítulo, recomendamos os pacotes [Numpy](https://numpy.org/) e [Matplotlib](https://matplotlib.org/). Tenha em mente que a função [sqrt](https://numpy.org/doc/stable/reference/generated/numpy.sqrt.html) do Numpy só retorna valores negativos se o valor de entrada tiver formato complex64, complex128, complex192 ou complex256.

## Exercício 1

**Passo 1:** Importar os pacotes necessários;

```python
import numpy as np
import matplotlib.pyplot as plt
```

**Passo 2:** Declare um vetor com as razões $z_2/z_1$;

```python
z2z1 = np.linspace(0, 2, 100)
```

**Passo 3:** Divida as frações por $z_1$ para achar $R$ e $T$ em função de $z_2/z_1$;

```python
R = (z2z1 - 1)/(z2z1 + 1)
T = 2*z2z1/(z2z1 + 1)
```

**Passo 4:** Plote o gráfico com as curvas.

```python
plt.plot(z2z1, R, 'r', linewidth=2)
plt.plot(z2z1, T, 'b', linewidth=2)
plt.legend(['Reflexão', 'Transmissão'])
plt.grid()
plt.xlabel('$z_2/z_1$')
plt.ylabel('Coeficiente')
```

## Exercício 2

**Passo 1:** Importar os pacotes necessários;

```python
import numpy as np
import matplotlib.pyplot as plt
```

**Passo 2:** Definir as variáveis necessárias, junto de um vetor de ângulos de incidência $\theta$ entre 0º e 90°;

```python
rho1 = np.complex128(1000.)
c1 = np.complex128(1480.)
rho2 = np.complex128(2700.)
c2 = np.complex128(6400.)

thetas = np.linspace(0, np.pi/2, 100, dtype=np.complex128)
thetas_graus = np.abs(thetas * 180 / np.pi)
```

**Passo 3:** Definir o cosseno de $\phi$ através dos ângulos de incidência $\theta$;

```python
cossenos_phi = np.sqrt(1 - (c2**2 * np.sin(thetas)**2)/c1**2)
```

**Passo 4:** Achar $R$ e $T$ em função dos ângulos de incidência;

```python
R = (rho2*c2*np.cos(thetas) - rho1*c1*cossenos_phi)/(rho2*c2*np.cos(thetas) + rho1*c1*cossenos_phi)
T = (2*rho2*c2*np.cos(thetas))/(rho2*c2*np.cos(thetas) + rho1*c1*cossenos_phi)
```

**Passo 5:** Plote os gráficos de magnitude e fase.

```python
fig, axes = plt.subplots(2, 1)
axes[0].plot(thetas_graus, np.abs(R), 'r', linewidth=2)
axes[0].plot(thetas_graus, np.abs(T), 'b', linewidth=2)
plt.grid()
axes[0].set_ylabel('Magnitude')

axes[1].plot(thetas_graus, np.angle(R), 'r', linewidth=2)
axes[1].plot(thetas_graus, np.angle(T), 'b', linewidth=2)
plt.grid()
axes[1].set_ylabel('Fase [rad]')

plt.xlabel('Ângulo de Incidência [°]')
```
